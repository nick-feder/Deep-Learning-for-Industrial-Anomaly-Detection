```python
import torch
from rd4ad_test import visualization, test
import os
import shutil

%load_ext autoreload
%autoreload 2
def test_paperclips(class_name="paperclips_front", vis=False):
    os.makedirs('./checkpoints', exist_ok=True)
    expected_path = f'./checkpoints/rm_1105_wres50_ff_mm_{class_name}.pth'
    if not os.path.exists(expected_path):
        source_path = f'C:\\Thesis\\Models\\rd4ad\\checkpoints\\wres50_{class_name}_best.pth'
        shutil.copy2(source_path, expected_path)
        print(f"Copied model from {source_path} to {expected_path}")
    
    print("Running test function...")
    auroc = test(class_name)
    print(f"Test AUROC: {auroc}")
    if vis == True:
        print("\nRunning visualization function...")
        visualization(class_name)
```

```python
test_paperclips('paperclips_front')
```
