```python
%load_ext autoreload
%autoreload 2
import os
import torch
from types import SimpleNamespace
from main import train, evaluate_with_visualization
base_results_dir ="./results"
base_checkpoint_dir = "./checkpoints"

views = ['paperclips_front', 'paperclips_side']
train_model = False  

for view in views:

    args = SimpleNamespace()
    args.config = "./fastflow_config.yaml"
    args.data = "../../data/paperclips/"
    args.category = view
    args.eval = not train_model
    args.checkpoint = os.path.join(base_checkpoint_dir, view)
    save_dir = os.path.join(base_results_dir, view)
    os.makedirs(save_dir, exist_ok=True)
    os.makedirs(args.checkpoint, exist_ok=True)
    
    if not args.eval:
        print(f"Training model for {view}")
        model = train(args)
        torch.save(model.state_dict(), os.path.join(args.checkpoint, f"model_{view}.pth"))
    print(f"Evaluating {view}")
    results = evaluate_with_visualization(args, save_dir=save_dir)
    print(f"{view} - Image AUROC: {results['image_auroc']:.4f}, Pixel AUROC: {results['pixel_auroc']:.4f}")
        
print("All views processed")
```
